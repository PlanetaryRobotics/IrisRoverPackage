```python
import time
def get_data_fake(ms_per_packet: int):
    """
    Fakes a packet stream by returning the first `n` packets from a list of 
    packets where `n` is determined by the number of milliseconds elapsed since 
    the first time this function was called and the rate at which packets should 
    arrive (number of milliseconds between each packet). 

    Args:
        ms_per_packet (int): Number of milliseconds elapsed between each packet.
    """
    # Initialize on first run:
    if get_data_fake.start_time < 0:
        get_data_fake.start_time = time.time()
    
    # Figure out how many packets should have arrived:
    ms_elapsed = (time.time() - get_data_fake.start_time) * 1000.0
    n_packets = int(ms_elapsed / ms_per_packet)
    
    # Return that many packets:
    packets_received = pcap_data['extracted_packets'][:n_packets]
    return packets_received
get_data_fake.start_time = -1
```
